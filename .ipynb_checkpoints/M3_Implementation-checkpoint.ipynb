This project analyzes currency exchange data to identify:

## Setup and Libraries

We import libraries for:

```python
#read data
df = pd.read_csv("Mercenaries.currency.csv", sep=";")
df.head()
df.shape
df.info()
df.describe()

#used for things later these are my column names
df.columns = [
    "League",
    "Date",
    "Get",
    "Pay",
    "Value",
    "Confidence"
]
```

## Data Filtering and Preprocessing

In this step, the dataset is cleaned and prepared for analysis. Since multiple currencies exist in the dataset, only transactions using **Chaos Orbs** are retained. Chaos Orbs are commonly considered the standard trading currency, making them ideal for normalization and consistent comparison.

The following preprocessing steps are performed:
- Filter the dataset to include only rows where the payment type is Chaos Orbs
- Convert the `Date` column into a proper datetime format for accurate sorting and grouping
- Extract time-based features such as **day**, **month**, and **week** to enable temporal analysis
- Compute the average exchange rate for each currency relative to Chaos Orbs

```python
#Data-set filtering
df = df[df["Pay"] == "Chaos Orb"] #only using rows that are priced in chaos for normalization 
df.head()
len(df) #26906 rows priced in chaos

df["Date"] = pd.to_datetime(df["Date"]) #better for date storting

df.groupby("Get")["Value"].mean().sort_values(ascending=False) #Average exchange rate per currency

#date descriptions to group based on weeks/months and later early late league
df["Date"] = pd.to_datetime(df["Date"])
df["Day"] = df["Date"].dt.day
df["Month"] = df["Date"].dt.month
df["Week"] = df["Date"].dt.isocalendar().week
```

# Discovery Question 2: Early vs Late League Trends

This section compares currency values between the early and late stages of the league.

The analysis highlights:

# Discovery Question 3: Outliers and K-Means Clustering

## Optimal Number of Clusters (Elbow Method)

The elbow method is used to determine the optimal number of clusters for K-Means by analyzing how inertia changes as the number of clusters increases.

The optimal value of *k* is chosen at the point where the rate of decrease in inertia begins to slow, indicating diminishing returns from adding more clusters.

```python
features = df[["Week", "Value"]].copy()

scaler = StandardScaler()
weekValue = scaler.fit_transform(features)

inertia = []
K = range(1, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(weekValue)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(6,4))
plt.plot(K, inertia, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method (Week vs Value)")
plt.show()
```

Here we can see that the sharp drop of intertia stops at 3 so we set our clustering to 3. 

```python
from sklearn.metrics import silhouette_score

sil_scores = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(weekValue)
    sil_scores.append(silhouette_score(weekValue, labels))

plt.figure(figsize=(6,4))
plt.plot(K, sil_scores, marker='o')
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score (Early vs Late League Values)")
plt.show()

best_k = K[sil_scores.index(max(sil_scores))]
print("Best k based on silhouette:", best_k)
```

## K-Means Behavioral Clustering

This section analyzes in-game items from Path of Exile using engineered market features such as price trend, volatility, average value, and peak timing across league weeks.

After standardizing these features, K-Means clustering (K = 3) is applied to group items with similar economic behavior patterns.

The resulting visualizations reveal how items separate into distinct market roles, including stable trade staples, high-volatility speculative items, and extreme outliers driven by spikes or late-league demand shifts.

Overall, this clustering helps compare how different items behave over time and how they relate to each other in terms of stability, growth, and market impact.

```python
features = []

for item, group in df.groupby("Get"):

    group = group.sort_values("Week")

    y = group["Value"].values.astype(float)
    x = group["Week"].values.astype(float)

    # slope (trend over time)
    if np.all(y == y[0]):
        slope = 0
    else:
        slope = np.polyfit(x, y, 1)[0]

    # volatility (stability of item price)
    mean_y = np.mean(y)
    volatility = np.std(y) / (mean_y + 1e-6)

    # overall economic tier
    mean_value = np.log1p(mean_y)

    # timing of peak value
    peak_ratio = np.argmax(y) / (len(y) - 1) if len(y) > 1 else 0

    features.append([
        item,
        slope,
        volatility,
        mean_value,
        peak_ratio
    ])

feature_df = pd.DataFrame(features, columns=[
    "Get",
    "slope",
    "volatility",
    "mean_value",
    "peak_ratio"
])

X = feature_df.drop(columns=["Get"])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(
    n_clusters=3,
    init="k-means++",
    n_init=20,
    max_iter=300,
    random_state=42
)

feature_df["Cluster"] = kmeans.fit_predict(X_scaled)


cluster_map = feature_df.set_index("Get")["Cluster"]

df["Cluster"] = df["Get"].map(cluster_map)


sns.scatterplot(
    data=feature_df,
    x="slope",
    y="volatility",
    hue="Cluster",
    palette="Set1"
)

plt.title("PoE Economy Clusters (K=3)")
plt.xlabel("Price Trend (Slope)")
plt.ylabel("Volatility")
plt.show()

sns.scatterplot(
    data=df,
    x="Value",
    y="Week",
    hue="Cluster",
    palette="Set1",
    alpha=0.6
)

plt.title("Value vs Week (Clustered Items)")
plt.xlabel("Value")
plt.ylabel("Week")
plt.show()
```

### Clustering Volatility vs Price Trend (Slope) 
A linear regression is applied to model value over time (Week vs Value).
- Positive slope → increasing value over time
- Negative slope → decreasing value
- Zero slope → stable pricing

Volatility (Price Stability) is calculated as the normalized standard deviation of values:
- High volatility → large price swings
- Low volatility → stable pricing

Mean Value (Economic Tier) is the average value log-transformed using log1p to reduce the effect of extreme values.
- Helps distinguish high-tier vs low-tier currencies
  
Peak Ratio (Timing of Peak Price) measures when the item reaches its maximum value during the observed time period:
- Close to 0 → peaks early
- Close to 1 → peaks late

We can see that most values grow or shrink at relatively the same speed across the length of the league, however many items have relatively high volatility. 

### Value vs Week (Clustered Items)
We can see that generally three major clusters appear: 
- A cluster of low-value items.
- A cluster of medium-value items.
- A cluster of high-value items that continue to trend upwards. 

## Hierarchical Clustering: Early vs Late Behavior

This section analyzes selected currencies using engineered features such as early value, late value, percentage change, and volatility.

After scaling the features, agglomerative (hierarchical) clustering is applied to group currencies with similar long-term behavior.

The resulting visualization helps compare how different currencies shift between early and late league stages and how they relate to each other in terms of growth and stability.

```python
df["Value_z"] = zscore(df["Value"])
# Outliers beyond ±3 std deviations
outliers = df[abs(df["Value_z"]) > 3]

overall_mean = df["Value"].mean()
currency_mean = df.groupby("Get")["Value"].mean()
currency_deviation = (currency_mean - overall_mean).abs().sort_values(ascending=False)
currency_deviation.head(10)

outliers["Get"].value_counts().head(10)
currency_deviation.head(10).plot(kind="bar")
plt.title("Currencies Deviating Most from Market Average")
plt.ylabel("Absolute Deviation from Mean")
plt.show()


kmeans = KMeans(n_clusters=3, random_state=42)
df["Cluster"] = kmeans.fit_predict(weekValue)

#cluster 2
from sklearn.cluster import DBSCAN

currency_stage_whitelist = currency_stage.loc[whitelist].copy()

features = pd.DataFrame({
    "Early": currency_stage_whitelist["Early"],
    "Late": currency_stage_whitelist["Late"],
    "PctChange": (currency_stage_whitelist["Late"] - currency_stage_whitelist["Early"]) / currency_stage_whitelist["Early"],
    "Volatility": df[df["Get"].isin(whitelist)].groupby("Get")["Value"].std()
}).fillna(0)

features.index.name = "Currency"
features.head()

features_log = features[["Early","Late","Volatility"]].apply(np.log1p)
features_log["PctChange"] = features["PctChange"]

# Scale to mean=0, std=1
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features_log)
features_scaled = pd.DataFrame(scaled_features, columns=features_log.columns, index=features_log.index)

agg_model = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
features_scaled["AggCluster"] = agg_model.fit_predict(features_scaled)

plt.figure(figsize=(10,6))
sns.scatterplot(data=features_scaled, x="Early", y="Late", hue="AggCluster", palette="Set2", s=100)
for i in range(len(features_scaled)):
    plt.text(features_scaled["Early"][i]+0.02, features_scaled["Late"][i]+0.02, features_scaled.index[i], fontsize=8)
plt.xlabel("Early Value (log-scaled & standardized)")
plt.ylabel("Late Value (log-scaled & standardized)")
plt.title("Agglomerative Clustering: Early vs Late Values")
plt.legend(title="Cluster")
plt.show()
```

### Currency Value Deviation
We can see that the Mirror of Kalandra has by far the greatest value. Furthermore, the Mirror Shard (20 of which create a Mirror of Kalandra) is the 3rd most valuable currency, demonstrating that even 1/20th of the Mirror's value is still more valuable than everything in the game with one exception. 

### Agglomerative Clustering: Early vs Late Values
Here we can see that from the beginning of the league to the end of the league the Mirror maintains its value. 

## PCA Visualization of Currency Trends

This section applies Principal Component Analysis (PCA) to the time-series data of currency values across weeks.

Each currency is represented by its value over time (week vs value), and PCA is used to reduce this high-dimensional data into two principal components. This allows the overall behavior of each currency to be visualized in a 2D space.

Currencies that appear close together on the plot exhibit similar trends over time, while those farther apart behave differently. This helps reveal underlying patterns in growth, decline, and volatility, and provides a visual foundation for clustering and further analysis.

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

scaler = StandardScaler()
scaled = scaler.fit_transform(pivot_white)

pca = PCA(n_components=2)
components = pca.fit_transform(scaled)

plt.figure()

plt.scatter(
    components[:, 0],
    components[:, 1]
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of Currency Value Trends (Week vs Value)")

plt.show()
```

### Interpretation of Principal Components

- **PC1 (Principal Component 1):** Represents the primary trend in the data. In this context, it typically reflects the overall direction of currency values over time, such as general growth or decline from early to later weeks.

- **PC2 (Principal Component 2):** Captures secondary patterns not explained by PC1. This often relates to differences in volatility or fluctuation, distinguishing stable currencies from those with more irregular or spiking behavior.

Together, PC1 and PC2 summarize the most important variations in the dataset, allowing complex time-series patterns to be visualized and compared in a simplified 2D space.
